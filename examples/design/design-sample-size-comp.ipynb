# Design > Sample Size Comparison

<div class="alert alert-info">Calculate the required sample size for comparing two means or proportions</div>

The `sample_size_comp` function determines sample size requirements for A/B tests and other two-group comparisons. You can solve for any one of sample size, power, confidence level, effect size, or (for means) the standard deviation, by leaving that parameter as `None`.

```python
import math

import pyrsm as rsm
```

# Example: Comparing two proportions

You are running an A/B test for a website. The current conversion rate is 1.0% (p1 = 0.01). You want to detect if a new design reduces the conversion rate to 0.8% (p2 = 0.008). You want 90% power at a 95% confidence level, using a one-sided (less than) test.

How many visitors do you need in each group?

```python
ssc = rsm.design.sample_size_comp(
    "proportion",
    p1=0.008,
    p2=0.01,
    conf=0.95,
    power=0.9,
    alt_hyp="less",
)
ssc.summary()
```

You need approximately **38,073** visitors per group (about 76,146 total) to reliably detect a difference this small. Small effect sizes require large sample sizes.

# Solving for power

If you already know your sample sizes and want to find out how much power you have, leave `power` as `None`:

```python
ssc_power = rsm.design.sample_size_comp(
    "proportion",
    n1=38073,
    n2=38073,
    p1=0.008,
    p2=0.01,
    conf=0.95,
    alt_hyp="less",
)
ssc_power.summary()
```

As expected, with 38,073 per group we get approximately 90% power.

# Example: Comparing two means

You want to test whether a new training program improves test scores by at least 5 points. The standard deviation of test scores is 10. You want 80% power with a two-sided test at 95% confidence.

```python
ssc_mean = rsm.design.sample_size_comp(
    "mean",
    delta=5,
    sd=10,
    conf=0.95,
    power=0.8,
    alt_hyp="two-sided",
)
ssc_mean.summary()
```

# Power curve

The `plot()` method generates a power curve showing how power changes with sample size. The dashed red line marks the calculated sample size:

```python
ssc_mean.plot()
```

# Solving for confidence level

Given fixed sample sizes and desired power, you can also solve for the required confidence level by leaving `conf` as `None`:

```python
ssc_conf = rsm.design.sample_size_comp(
    "proportion",
    n1=38073,
    n2=38073,
    p1=0.008,
    p2=0.01,
    power=0.9,
    alt_hyp="less",
)
ssc_conf.summary()
```

# Two-sided proportion test

For a two-sided test comparing conversion rates of 10% vs 15%:

```python
ssc_two = rsm.design.sample_size_comp(
    "proportion",
    p1=0.1,
    p2=0.15,
    conf=0.95,
    power=0.8,
    alt_hyp="two-sided",
)
ssc_two.summary()
```
