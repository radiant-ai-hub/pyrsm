# State > Save and Load State

<div class="alert alert-info">Persist selected Python objects so a notebook can resume without recomputing every step.</div>


<!-- pyrsm-teaching-note: state-roadmap -->
## Teaching Roadmap

Use this notebook to teach reproducibility and workflow management. Students should understand what is saved, where it is saved, why some Python objects cannot be pickled, and why it is usually better to save a small dictionary of needed objects than the entire global environment.


```python
x = 2
y = 3
z = x - y
```

```python
import pyrsm as rsm
```

<!-- pyrsm-teaching-note: state-save-globals -->
### Saving the current environment

`globals()` contains all objects currently defined in the notebook session. Saving it is convenient for demonstrations, but it can also capture temporary objects that are not needed later. For production work, prefer saving the specific objects required to continue the analysis.


```python
rsm.notebook.save_state(globals(), "save-load.state.pkl")
```

```python
del x, y, z
```

```python
try:
    print(x)
except Exception as err:
    print(err)
```

Use `rsm.notebook.load_state` to either return a dictionary of the objects you stored, or add them to `globals()` for easy access

<!-- pyrsm-teaching-note: state-load-globals -->
### Loading back into the notebook

Passing `globals()` to `load_state` restores the saved names directly into the active notebook environment. That is useful when continuing an analysis, but students should recognize that it changes the session state rather than returning a visible table.


```python
rsm.notebook.load_state("save-load.state.pkl", globals())
```

```python
x
```

If your notebook contains objects that cannot be pickled, you can save just the objects you want to keep as follows

```python
rsm.notebook.save_state({"x": x, "y": y, "z": z}, "objects-to-keep.state.pkl")
```

```python
del x, y, z
```

```python
try:
    print(x)
except Exception as err:
    print(err)
```

```python
rsm.notebook.load_state("objects-to-keep.state.pkl")
```

<!-- pyrsm-teaching-note: state-returned-dict -->
### Inspecting a returned state dictionary

When `load_state` is called without `globals()`, it returns a dictionary. This is often safer because students can inspect the restored object names before deciding what to put back into the notebook environment.


```python
restored = rsm.notebook.load_state("objects-to-keep.state.pkl")
list(restored.keys())
```

```python
# cleaning up ...
import os

os.remove("save-load.state.pkl")
os.remove("objects-to-keep.state.pkl")
```
