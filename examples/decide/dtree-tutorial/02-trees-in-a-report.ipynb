## Example 2

> You are an actor and would like to sign a contract with either a movie company or a TV network. If you sign with a movie company, you need to pay $5,000 for the legal fees. There will be a 30% chance of having small box office and earning $200,000, a 60% chance of having medium box office with payoff equal to $1,000,000, and a 10% chance of having large box office and getting $3,000,000. If you decide to sign the contract with TV networks instead, your payoff will be $900,000. What is EMV of signing the contract with movie company? What is your optimal strategy?

````python
import pyrsm as rsm
from IPython.display import Markdown


def show(tree, final=True):
    """Draw a decision tree (Mermaid; rendered by JupyterLab, VS Code, Quarto)."""
    return Markdown(f"```mermaid\n{tree.to_mermaid(final=final)}\n```")
````

A `variables:` section gives a value a name that the tree can use, here the legal fees. The fees are a `cost` of signing with the movie company.

````python
sign_contract = """name: Sign contract
variables:
    legal fees: 5000
type: decision
Sign with Movie Company:
    cost: legal fees
    type: chance
    Small Box Office:
        p: 0.3
        payoff: 200000
    Medium Box Office:
        p: 0.6
        payoff: 1000000
    Large Box Office:
        p: 0.1
        payoff: 3000000
Sign with TV Network:
    payoff: 900000
"""
````

````python
tree1 = rsm.decide.dtree(sign_contract)
tree1.summary(input=False, output=True)
````

The EMV of signing with the movie company is 0.3 × 200,000 + 0.6 × 1,000,000 + 0.1 × 3,000,000 − 5,000 = **955,000**, more than the 900,000 from the TV network: **sign with the movie company**.

````python
show(tree1)
````

### What if the legal fees are $10,000?

````python
fees_10000 = """name: Sign contract
variables:
    legal fees: 10000
type: decision
Sign with Movie Company:
    cost: legal fees
    type: chance
    Small Box Office:
        p: 0.3
        payoff: 200000
    Medium Box Office:
        p: 0.6
        payoff: 1000000
    Large Box Office:
        p: 0.1
        payoff: 3000000
Sign with TV Network:
    payoff: 900000
"""
````

````python
tree2 = rsm.decide.dtree(fees_10000)
tree2.summary(input=False, output=True)
````

The EMV drops to $950,000: still sign with the movie company.

### What if a medium box office pays $800,000?

````python
medium_800000 = """name: Sign contract
variables:
    legal fees: 5000
type: decision
Sign with Movie Company:
    cost: legal fees
    type: chance
    Small Box Office:
        p: 0.3
        payoff: 200000
    Medium Box Office:
        p: 0.6
        payoff: 800000
    Large Box Office:
        p: 0.1
        payoff: 3000000
Sign with TV Network:
    payoff: 900000
"""
````

````python
tree3 = rsm.decide.dtree(medium_800000)
tree3.summary(input=False, output=True)
````

The movie company is now worth 0.3 × 200,000 + 0.6 × 800,000 + 0.1 × 3,000,000 − 5,000 = 835,000, less than $900,000: **sign with the TV network**.

````python
show(tree3)
````
